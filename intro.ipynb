```python
# Concept of indexing and slicing in Python

str="Hello World"
print(str[3::])
print(str[6::])
print(str[3:8])
print(str[::2])
print(str[::-1])
```

```python
# Palindrome check in Python    
str_1=input("Enter a string : ")
str_2=str_1[::-1]
if str_1==str_2:
    print("the string is palindrome")
else:
    print("the string is not palindrome")
```

```python
# logical operators with if else statement

'''
if(conditon):
    statement1
elif(condition):
    statement2
else:
    statement3
'''
Marks=int(input("Enter marks of a Students :"))
if (Marks>=90):
    print("Grade A+")
elif(Marks>=80):
    print("Grade A")
elif(Marks>=70):
    print("Grade B+")
elif(Marks>=60):
    print("Grade B")
else:
    print("Grade D")
```
